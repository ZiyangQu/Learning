44.假设一个 $n×(n+1)$ 的矩阵被行化简为简化行阶梯形矩阵。当n=30 时，当n=300 时，化简的向后阶段所涉及的运算次数大约占总运算次数的多少比例？  
答：前向消元将主元下面的元素消为零。对于第 $k$ 列，需要处理 $n-k$ 行，每行更新 $n-k+1$ 个元素，每个元素约需要 2 次浮点运算(一次乘法一次加法)。因此前向消元的运算量为：
$$
F=2\sum_{k=1}^{n-1}(n-k)(n-k+1)
$$
化简并保留最高阶项：
$$
\boxed{F\approx\frac{2}{3}n^3}
$$
后向消元的目的是将主元上面的元素消为零。
对于第 $k$ 列，需要处理 $k-1$ 行，每行更新 $n-k+1$ 个元素。
因此后向消元的运算量为：
$$
B=2\sum_{k=2}^{n}(k-1)(n-k+1)
$$
化简并保留最高阶项：
$$
\boxed{B\approx\frac{1}{3}n^3}
$$
总运算量为：
$$
F+B\approx n^3
$$
因此后向消元所占比例为：
$$
\frac{B}{F+B}\approx\frac{1}{3}
=33.3\%
$$
当 $n=30$ 时，后向消元约占总运算量的 $33.3\%$。
当 $n=300$ 时，后向消元约占总运算量的 $33.3\%$。
两种情况下，后向消元均约占总浮点运算量的三分之一。

```python
#求解插值多项式并计算阻力
import numpy as np

# 实验数据
t = np.array([0, 2, 4, 6, 8, 10], dtype=float)
F = np.array([0, 2.90, 14.8, 39.6, 74.3, 119], dtype=float)

# 构造范德蒙德矩阵:把多项式中的各项系数组织成一个线性方程组。
A = np.vander(t, 6, increasing=True)

# 求解线性方程组
a = np.linalg.solve(A, F)

# 输出多项式系数
print("五次插值多项式系数：")
for i, ai in enumerate(a):
    print(f"a{i} = {ai:.8f}")

# 计算750 ft/sec时的阻力
t_test = 7.5
force = np.polynomial.polynomial.polyval(t_test, a)

print("\n750 ft/sec时的阻力：")
print(f"{force:.5f} × 100 lb")
print(f"实际阻力 = {force * 100:.2f} lb")
```

证明：
因为 $w$ 是方程组 $Ax=b$ 的任意一个解，所以：
$$
Aw=b
$$
又因为 $p$ 是方程组 $Ax=b$ 的一个特解，所以：
$$
Ap=b
$$
两式相减，得到：
$$
Aw-Ap=b-b=0
$$
根据矩阵乘法的线性性质：
$$
A(w-p)=0
$$
又因为：
$$
v_h=w-p
$$
所以：
$$
Av_h=0
$$
因此，$v_h$ 是齐次方程组 $Ax=0$ 的一个解。
由定义式：
$$
v_h=w-p
$$
移项得到：
$$
w=p+v_h
$$
证毕。
因此，非齐次线性方程组 $Ax=b$ 的任意一个解，都可以表示为该方程组的一个特解与对应齐次方程组 $Ax=0$ 的一个解之和。

由于最后一行没有主元位置，因此矩阵的列向量不能张成整个 $\mathbb{R}^5$。

所以：

$$

$$
第47、48题均存在没有主元位置的行，因此两个线性变换都不是满射。


```python

import sympy as sp

# 输入第47题和第48题的矩阵
A47 = sp.Matrix([
    [4, -7, 3, 7, 5],
    [6, -8, 5, 12, -8],
    [-7, 10, -8, -9, 14],
    [3, -5, 4, 2, -6],
    [-5, 6, -6, -7, 3]
])

A48 = sp.Matrix([
    [9, 13, 5, 6, -1],
    [14, 15, -7, -6, 4],
    [-8, -9, 12, -5, -9],
    [-5, -6, -8, 9, 8],
    [13, 14, 15, 2, 11]
])

matrices = {47: A47, 48: A48}

for num, A in matrices.items():

    # 求简化阶梯形矩阵
    R, pivots = A.rref()

    print(f"\n第{num}题")
    print("简化阶梯形矩阵：")
    sp.pprint(R)

    # 检查每一行是否都有主元位置
    every_row_has_pivot = all(
        any(R[i, j] != 0 for j in range(A.cols))
        for i in range(A.rows)
    )

    if every_row_has_pivot:
        print("每一行都有主元位置")
        print("T 是满射（Onto）")
    else:
        print("存在没有主元位置的行")
        print("T 不是满射（Not Onto）")

    print("-" * 45)
```

根据基尔霍夫电压定律（KVL），分别列出四个网孔的电压方程。

第一个网孔：

3I_1+2I_1+(I_1-I_2)=-30

整理得：

6I_1-I_2=-30

第二个网孔：

整理得：

第三个网孔：

第四个网孔：

$$

$$

建立矩阵方程：

将上述四个方程写成矩阵形式：

$$
\begin{bmatrix}
6&-1&0&0\\
-1&9&-4&0\\
0&-4&7&-2\\
0&0&-2&7
\end{bmatrix}
\begin{bmatrix}
I_1\\I_2\\I_3\\I_4

=

求解结果：

利用 Python 求解矩阵方程，得到四个回路电流。

$$
\begin{aligned}
I_1&\approx-5.0496\text{ A}\\
I_2&\approx-0.2977\text{ A}\\
I_3&\approx5.5925\text{ A}\\
I_4&\approx0.1693\text{ A}
\end{aligned}
$$

其中，负号表示实际电流方向与假设的顺时针方向相反。


```python

import sympy as sp

# 定义电阻系数矩阵
A = sp.Matrix([
    [6, -1,  0,  0],
    [-1, 9, -4,  0],
    [0, -4,  7, -2],
    [0,  0, -2,  7]
])

# 定义电源电压向量
b = sp.Matrix([
    -30,
    -20,
    40,
    -10
])

# 求解矩阵方程 AI = b
I = A.inv() * b

# 也可以使用 LUsolve 求解
I = A.LUsolve(b)

# 输出矩阵方程
print("电阻系数矩阵 A：")
sp.pprint(A)

print("\n电源电压向量 b：")
sp.pprint(b)

# 输出精确解
print("\n回路电流精确值：")
sp.pprint(I)

# 输出小数结果
print("\n回路电流数值（A）：")
for i, value in enumerate(I):
    print(f"I{i+1} = {float(value):.4f} A")

# 验证计算结果
print("\n验证 AI = b：")
sp.pprint(A * I)

print("\n验证是否成立：", A * I == b)
```
